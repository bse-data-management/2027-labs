## Part C — Send malformed requests (8 min, optional)

Each cell sends one request the API should refuse, and prints the status and the
body. They work before part B is finished: the refusal happens before your code
runs. You can also paste the paths after `http://localhost:8000` in a browser.

For each, note the status, and which parameter and which rule refused it. The
body's `loc` and `type` say.

```python
import json

import httpx


def call(path: str) -> None:
    """Send GET path to the API, and print the status and the body."""
    response = httpx.get(f"http://localhost:8000{path}")
    print(response.status_code, response.reason_phrase)
    print(json.dumps(response.json(), indent=2))
```

```python
call("/drivers/abc")
```

```python
call("/drivers/42/earnings?to=2024-01-14")
```

```python
call("/drivers/42/earnings?from=08/01/2024&to=2024-01-14")
```
